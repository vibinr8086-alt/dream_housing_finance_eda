### Data Cleaning and Validation

The dataset was checked for missing values, duplicate records, data types and potential outliers.

- Missing values were treated using mode imputation for categorical variables and median imputation for LoanAmount.
- Loan_Amount_Term was filled using its mode (360 months).
- Credit_History was filled using its mode (1.0), with the limitation that this may introduce bias.
- No duplicate rows were identified.
- Potential outliers in ApplicantIncome, CoapplicantIncome and LoanAmount were retained because they may represent genuine customer records.
- Data types were inspected and considered appropriate for the analysis.

After cleaning, the dataset contained 614 rows and 13 columns, with no missing values or duplicate rows.

# Exploratory Data Analysis (EDA) Report

## Dream Housing Finance – Loan Eligibility Prediction

Dataset: 614 loan applications | Target variable: Loan_Status | Analysis completed: Univariate, bivariate and multivariate analysis

## 1. Project objective

The objective of this EDA is to understand the factors associated with loan approval and rejection. We explored applicant demographics, employment status, income, loan amount, credit history and property area to identify patterns and relationships with loan status.

## 2. Dataset overview

- Total records: 614 loan applications.
- Each row represents one applicant's loan application.
- Target variable: `Loan_Status`.
- `Y` represents loan approved.
- `N` represents loan rejected.
- The dataset contains categorical and numerical variables.

## 3. Overall loan approval distribution

Approved (Y)

Rejected (N)

- 422 applications were approved (approximately 68.73%).
- 192 applications were rejected (approximately 31.27%).
- The dataset contains more approved applications than rejected applications.

Insight: Loan approvals represent the majority of applications. A future classification model should be evaluated carefully to ensure it can identify rejected applications as well as approved ones.

## 4. Univariate analysis – Individual variable analysis

Univariate analysis helped us understand the distribution of each variable independently.

| Variable          | Key observation                                                        |
| ----------------- | ---------------------------------------------------------------------- |
| Gender            | Male applicants are considerably more numerous than female applicants. |
| Married           | Married applicants form the larger group.                              |

Main insight: The dataset is dominated by certain applicant groups, particularly male, married, graduate and non-self-employed applicants. Income and loan amount distributions also contain outliers.

We compared individual variables with `Loan_Status` to understand how applicant characteristics relate to loan approval.

| Variable          | Key observation                                                                             |

## 9. Overall key findings

The following are the most important conclusions from the complete EDA:

1\. Credit history is strongly associated with loan approval. Applicants with a positive credit-history indicator had a much higher approval proportion.

2\. Applicant income alone does not determine loan approval. Both higher- and lower-income applicants appear in approved and rejected groups.

3\. Loan amount is not a decisive factor by itself. Approved and rejected applications have overlapping loan-amount distributions.

4\. Applicant income and loan amount have a moderate positive correlation. Applicants with higher incomes tend to request larger loans.

5\. Demographic variables show overlapping outcomes. Gender, marital status, education and employment status do not independently provide a clear separation between approved and rejected applications in the analyses performed.

6\. The dataset contains outliers and skewed numerical distributions. These should be considered during preprocessing and model selection.

7\. Missing values require appropriate treatment. Credit_History has the largest missing-value percentage and deserves particular attention.

## 10. Final EDA conclusion

The exploratory analysis of the Dream Housing Finance dataset reveals that credit history has a strong association with loan approval, whereas applicant income and loan amount alone do not clearly determine the outcome. Demographic characteristics also show overlapping approval and rejection patterns. The numerical variables contain skewness and outliers, and several columns contain missing values.

Overall, EDA provides a better understanding of the dataset and identifies important patterns that can guide data preprocessing, feature preparation and classification model development.
